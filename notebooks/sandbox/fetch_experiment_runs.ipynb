# `fetch_experiment_runs` — Function Tests

Tests all combinations of the `include_summary_metrics` parameter and state filtering (passed via the `filters` dict).
Each section runs one variant and prints shape + column info so the results can be compared.

```python
from src.utils.config import fetch_experiment_runs

# Fixed filters used across all tests — narrow to matrix_factorization runs only
BASE_FILTERS = {"config.model": "matrix_factorization"}
```

## Test 1 — Default behaviour (no state filter, no summary metrics)

Baseline: should behave identically to the old function.
Expected: returns all runs regardless of state, no `summary_metrics` column.

```python
default_runs = fetch_experiment_runs(BASE_FILTERS)

print(f"Shape: {default_runs.shape}")
print(f"Columns: {default_runs.columns}")
assert "summary_metrics" not in default_runs.columns, "summary_metrics should NOT be present in default mode"
print("✓ No summary_metrics column")
default_runs.head(3)
```

## Test 2 — State filter via filters dict (`"state": "finished"`)

Expected: returns fewer rows than Test 1 (only finished runs).
No `summary_metrics` column.

```python
finished_runs = fetch_experiment_runs({**BASE_FILTERS, "state": "finished"})

print(f"Shape: {finished_runs.shape}")
print(f"All runs:    {default_runs.shape[0]}")
print(f"Finished:    {finished_runs.shape[0]}")
print(f"Difference:  {default_runs.shape[0] - finished_runs.shape[0]} (non-finished runs filtered out)")
assert finished_runs.shape[0] <= default_runs.shape[0], "Finished runs should be <= all runs"
assert "summary_metrics" not in finished_runs.columns, "summary_metrics should NOT be present"
print("✓ Row count reduced or equal")
print("✓ No summary_metrics column")
finished_runs.head(3)
```

## Test 3 — Summary metrics only (no state filter)

Expected: returns all runs (same count as Test 1), with a `summary_metrics` column present.
Inspect the column contents to verify the JSON is parsed correctly.

```python
with_summary_runs = fetch_experiment_runs(BASE_FILTERS, include_summary_metrics=True)

print(f"Shape: {with_summary_runs.shape}")
assert with_summary_runs.shape[0] == default_runs.shape[0], "Row count should match default (no state filter)"
assert "epoch/test_recall@10" in with_summary_runs.columns, "epoch metrics should be present as flat columns"
print("✓ Row count matches default")
print("✓ epoch metric columns present")

# Show all epoch/* columns
epoch_cols = [c for c in with_summary_runs.columns if c.startswith("epoch/")]
print(f"\nEpoch metric columns ({len(epoch_cols)}): {epoch_cols}")

# Renamed metadata columns
for col in ["runtime", "step", "timestamp"]:
    present = col in with_summary_runs.columns
    print(f"  {col}: {'✓' if present else '✗'}")

with_summary_runs.head(3)
```

## Test 4 — State filter + `include_summary_metrics=True` (the model_based_parse_params call pattern)

This is the exact combination that `model_based_parse_params` will use in production.
Expected: same row count as Test 2, with `summary_metrics` present and non-null for all rows (finished runs should always have metrics).

```python
finished_with_summary = fetch_experiment_runs({**BASE_FILTERS, "state": "finished"}, include_summary_metrics=True)

print(f"Shape: {finished_with_summary.shape}")
assert finished_with_summary.shape[0] == finished_runs.shape[0], "Row count should match state='finished' result"
assert "epoch/test_recall@10" in finished_with_summary.columns, "epoch metrics should be present as flat columns"
print("✓ Row count matches finished-only result")
print("✓ epoch metric columns present")

null_count = finished_with_summary["epoch/test_recall@10"].null_count()
print(f"\nRuns with null epoch/test_recall@10: {null_count} / {finished_with_summary.shape[0]}")
if null_count > 0:
    print("⚠ Some finished runs are missing the target metric")
else:
    print("✓ All finished runs have epoch/test_recall@10 populated")
finished_with_summary.head(3)
```

```python
# Inspect a sample row — metric values and renamed metadata fields
finished_with_summary.select(
    ["id", "name", "runtime", "step", "timestamp", "epoch/test_recall@10", "epoch/test_recall@20"]
).head(3)
```

## Test 5 — Config filters combined with state filter

Verify that `config.*` filters and `state` filter work together correctly.
Use a narrow filter (specific embedding_dimension) to get a small, fast result.

```python
lower_embedding_runs = fetch_experiment_runs(
    {
        "config.model": "matrix_factorization",
        "config.embedding_dimension": 64,
        "state": "finished",
    },
    include_summary_metrics=True,
)

print(f"Shape: {lower_embedding_runs.shape}")
print(f"Unique embedding_dimension values: {lower_embedding_runs['embedding_dimension'].unique().to_list()}")
assert all(v == 64 for v in lower_embedding_runs["embedding_dimension"].to_list()), "All rows should have embedding_dimension=64"
print("✓ Config filter applied correctly alongside state filter")
lower_embedding_runs.head(3)
```
