# `fetch_experiment_runs` — Function Tests

Tests all combinations of the `include_summary_metrics` parameter and state filtering (passed via the `filters` dict).
Each section runs one variant and prints shape + column info so the results can be compared.

```python
from src.utils.config import fetch_experiment_runs

# Fixed filters used across all tests — narrow to matrix_factorization runs only
BASE_FILTERS = {"model": "matrix_factorization"}
```

## Test 1 — Default behaviour (no state filter, no summary metrics)

Baseline: should behave identically to the old function.
Expected: returns all runs regardless of state, no `summary_metrics` column.

```python
df_default = fetch_experiment_runs(BASE_FILTERS)

print(f"Shape: {df_default.shape}")
print(f"Columns: {df_default.columns}")
assert "summary_metrics" not in df_default.columns, "summary_metrics should NOT be present in default mode"
print("✓ No summary_metrics column")
df_default.head(3)
```

## Test 2 — State filter via filters dict (`"state": "finished"`)

Expected: returns fewer rows than Test 1 (only finished runs).
No `summary_metrics` column.

```python
df_finished = fetch_experiment_runs({**BASE_FILTERS, "state": "finished"})

print(f"Shape: {df_finished.shape}")
print(f"All runs:    {df_default.shape[0]}")
print(f"Finished:    {df_finished.shape[0]}")
print(f"Difference:  {df_default.shape[0] - df_finished.shape[0]} (non-finished runs filtered out)")
assert df_finished.shape[0] <= df_default.shape[0], "Finished runs should be <= all runs"
assert "summary_metrics" not in df_finished.columns, "summary_metrics should NOT be present"
print("✓ Row count reduced or equal")
print("✓ No summary_metrics column")
df_finished.head(3)
```

## Test 3 — Summary metrics only (no state filter)

Expected: returns all runs (same count as Test 1), with a `summary_metrics` column present.
Inspect the column contents to verify the JSON is parsed correctly.

```python
df_with_summary = fetch_experiment_runs(BASE_FILTERS, include_summary_metrics=True)

print(f"Shape: {df_with_summary.shape}")
assert df_with_summary.shape[0] == df_default.shape[0], "Row count should match default (no state filter)"
assert "summary_metrics" in df_with_summary.columns, "summary_metrics column should be present"
print("✓ Row count matches default")
print("✓ summary_metrics column present")

# Inspect non-null summary_metrics rows
non_null = df_with_summary.filter(df_with_summary["summary_metrics"].is_not_null())
print(f"\nRuns with summary_metrics populated: {len(non_null)} / {len(df_with_summary)}")

# Show the keys available in a sample summary_metrics dict
if len(non_null) > 0:
    sample = non_null["summary_metrics"][0]
    print(f"\nSample summary_metrics keys:\n{sorted(sample.keys())}")
df_with_summary.head(3)
```

## Test 4 — State filter + `include_summary_metrics=True` (the model_based_parse_params call pattern)

This is the exact combination that `model_based_parse_params` will use in production.
Expected: same row count as Test 2, with `summary_metrics` present and non-null for all rows (finished runs should always have metrics).

```python
df_full = fetch_experiment_runs({**BASE_FILTERS, "state": "finished"}, include_summary_metrics=True)

print(f"Shape: {df_full.shape}")
assert df_full.shape[0] == df_finished.shape[0], "Row count should match state='finished' result"
assert "summary_metrics" in df_full.columns, "summary_metrics column should be present"
print("✓ Row count matches finished-only result")
print("✓ summary_metrics column present")

non_null_full = df_full.filter(df_full["summary_metrics"].is_not_null())
null_count = df_full.shape[0] - len(non_null_full)
print(f"\nRuns with null summary_metrics: {null_count} / {df_full.shape[0]}")
if null_count > 0:
    print("⚠ Some finished runs have no summary_metrics — inspect these rows:")
    print(df_full.filter(df_full["summary_metrics"].is_null()).select(["id", "name"]))
else:
    print("✓ All finished runs have summary_metrics populated")
df_full.head(3)
```

## Test 5 — Config filters combined with state filter

Verify that `config.*` filters and `state` filter work together correctly.
Use a narrow filter (specific embedding_dimension) to get a small, fast result.

```python
df_narrow = fetch_experiment_runs(
    {"model": "matrix_factorization", "embedding_dimension": 64, "state": "finished"},
    include_summary_metrics=True,
)

print(f"Shape: {df_narrow.shape}")
print(f"Unique embedding_dimension values: {df_narrow['embedding_dimension'].unique().to_list()}")
assert all(v == 64 for v in df_narrow["embedding_dimension"].to_list()), "All rows should have embedding_dimension=64"
print("✓ Config filter applied correctly alongside state filter")
df_narrow.head(3)
```
